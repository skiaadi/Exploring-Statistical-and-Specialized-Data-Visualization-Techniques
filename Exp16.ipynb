#Data Visualization and Encoding

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
data = {
    'Days': ['Mon','Tue','Wed','Thu','Fri'],
    'Study_Hours':[2,3,2.5,4,3.5],
    'Marks':[50,60,55,70,65],
    'Attendance':[60,70,65,80,75],
    'Sleep_Hours':[6,7,6,8,7],
    'Assignments_Completed':[1,2,1,3,2]
}
df=pd.DataFrame(data)
df.head()
```

```python
plt.plot(df['Days'],df['Study_Hours'],marker='*')
plt.title("STudy Hours Trend")
plt.xlabel("Days")
plt.ylabel("Study Hours")
plt.show()
```

```python
plt.figure(figsize=(7,4))
plt.plot(df['Days'],df['Study_Hours'],marker='o',color='blue',label='Study Hours')
plt.plot(df['Days'],df['Marks'],marker='s',color='red',label='Marks')
plt.title("Study Hours and Marks Trend")
plt.xlabel("Days")
plt.ylabel("Value")
plt.legend()
plt.show()
```

```python
#Bar CHart
plt.bar(df['Days'],df['Marks'],color='lime')
plt.title("Marks Distribution")
plt.xlabel("Days")
plt.ylabel("Marks")
plt.grid()
plt.show()
```

```python
bars = plt.bar(df['Days'],df['Marks'],color='teal')
for bar in bars:
  y = bar.get_height()
  plt.text(bar.get_x()+bar.get_width()/2, y, str (y),
           ha='center',va='bottom')

plt.title("Marks Distribution")
plt.xlabel("Days")
plt.ylabel("Marks")
plt.grid(axis='y')
plt.show()
```

```python
#Histogram
data = {
    'Marks':[5,60,55,70,65,75,80,85,90,45,68,72,77,82,88]
}
df=pd.DataFrame(data)
plt.hist(df['Marks'],bins=3,color='teal',edgecolor='k')
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
#Scatterplot
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Day':['Mon','Tue','Wed','Thu','Fri','Sat','Sun','Mon','Tue','Wed'],
    'Region':['North','South','East','West','North','South','East','West','North','South'],
    'Sales':[200,220,250,270,300,350,400,280,260,290],
    'Profit':[20,25,30,35,40,45,50,60,80,40],
    'Customers':[30,35,40,45,50,65,80,55,48,52],
    'Category':['Electronics','Clothing','Electronics','Clothing','Electronics','Clothing','Electronics','Clothing','Electronics','Clothing']
}
df=pd.DataFrame(data)
plt.scatter(df['Day'],df['Sales'],color='k')
plt.show()
```

```python
#Conditional Scatterplot
df['Result']=['Fail','Pass','Fail','Pass','Pass']
colors = ['red' if r == 'Fail' else 'green' for r in df['Result']]
plt.scatter(df['Days'],df['Sales'],c=colors)
plt.show()
```
